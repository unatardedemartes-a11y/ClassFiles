```python
import numpy as np
import scipy.linalg as la

# 1. Definimos la matriz B obtenida en el paso anterior
B = np.array([[6, 3, 1],
              [1, 1, 3],
              [0, -2, 2]])

print("MATRIZ B")
print(B)
print()

# 2. Factorización LU
# Nota: scipy.linalg.lu devuelve P, L, U. (P es la matriz de permutación)
P, L, U = la.lu(B)
print("FACTORIZACIÓN LU")
print("L (Triangular inferior):\n", L)
print("U (Triangular superior):\n", U)
print()

# 3. Factorización QR
Q, R = la.qr(B)
print("FACTORIZACIÓN QR")
print("Q (Matriz ortogonal):\n", Q)
print("R (Triangular superior):\n", R)
print()

# 4. Factorización SVD
U_svd, S, Vt = la.svd(B)
# scipy devuelve los valores singulares (S) en un arreglo de 1D, lo convertimos a matriz diagonal
Sigma = np.zeros((3, 3))
np.fill_diagonal(Sigma, S)

print("DESCOMPOSICIÓN SVD")
print("U (Vectores singulares izquierdos):\n", U_svd)
print("Sigma (Matriz diagonal de valores singulares):\n", Sigma)
print("V^T (Vectores singulares derechos transpuestos):\n", Vt)
```

```python
#Matriz A
import numpy as np
import scipy.linalg as la

# Matriz B (diagonalmente dominante)
B = np.array([[1, 1, 3],
              [6, 3, 1],
              [1, 3, 1]])

print("Matriz B:\n", B)

# 1. Factorización LU
P, L, U = la.lu(B)
print("\n--- Factorización LU ---")
print("L:\n", L)
print("U:\n", U)

# 2. Factorización QR
Q, R = la.qr(B)
print("\n--- Factorización QR ---")
print("Q:\n", Q)
print("R:\n", R)

# 3. SVD (Descomposición en Valores Singulares)
U_svd, S, Vt = la.svd(B)
# La función devuelve los valores singulares (S) en un vector. Lo convertimos a matriz diagonal.
Sigma = np.zeros((3, 3))
np.fill_diagonal(Sigma, S)

print("\n--- Factorización SVD ---")
print("U:\n", U_svd)
print("Sigma:\n", Sigma)
print("V^T:\n", Vt)
```
