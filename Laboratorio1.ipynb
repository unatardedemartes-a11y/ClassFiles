# Objetivos

```python
valores = []
# Abre el archivo en modo lectura ('r')
with open('/content/drive/MyDrive/ClassFiles/mbox-short.txt', 'r') as archivo:
  # Cursor (si le puedo decir asi? :D) para recorrer lineas
  for linea in archivo:
      # Si la linea inicia con el valor que requerimos entonces continua con extrar el numero
      if linea.startswith('X-DSPAM-Confidence'):
              # Extraer el numero despues de los dos puntos porque "X-DSPAM-Confidence" es un formato que siempre tiene la estructura X-DSPAM-Confidence: 0.8475
              numeros = linea.split(':')[1].strip()
              #print(numeros)
              # Guardar en la lista
              valores.append(numeros)

# Mostrar los resultados
print("RESULTADOS DE MEDICIONES X-DSPAM-Confidence")
for i in range(len(valores)):
    print("Indice {i}",i,":", valores[i])

conteo = len(valores)  #es extrano no usar count o solo me lo parece a mi?
print("Conteo:", conteo)

suma = sum([float(x) for x in valores])  #cuando se leyo la linea, se guardo como string, se conviernte a numerico para realizar operaciones
print("Suma:", suma)

promedio = suma / conteo
print("Promedio:", promedio)

# Sumar los cuadrados de las diferencias
suma_cuadrados = 0
for i in valores:
    valor_numerico = float(i)  #cuando se leyo la linea, se guardo como string, se conviernte a numerico para realizar operaciones
    suma_cuadrados += (valor_numerico - promedio) ** 2

# Varianza poblacional (dividir entre n)
varianza = suma_cuadrados / len(valores)
print("Varianza:", varianza)
```
