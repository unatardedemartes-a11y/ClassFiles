### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# Escribe tu código aquí
dup_first= hoteles.duplicated(keep='first').sum() #marca el siguiente registro que esta duplicado como false
dup_last= hoteles.duplicated(keep='last').sum()   #marca el ultimo registro que esta duplicado como false
dup_false= hoteles.duplicated(keep=False).sum()   #marca todos los registros que tienen una duplicidad
print("Duplicados first:",dup_first)
print("Duplicados last:",dup_last)
print("Duplicados false:",dup_false)
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Escribe tu código aquí
total_filas = len(hoteles)
porcentaje_sobrantes = (dup_first / total_filas) * 100  #porcentaje de filas sobrantes respecto al total de filas
print("Porcentaje de filas sobrantes:", round(porcentaje_sobrantes,2),"%")

grupos_repetidos = (dup_false - dup_first) // 2  #numero de grupos repetidos (combinaciones de valores que aparecen más de una vez
print("Número de grupos repetidos:", grupos_repetidos)
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Escribe tu código aquí

hoteles_dup = hoteles.copy() #hacer copia del dataset

hoteles_dup['es_duplicado'] = hoteles_dup.duplicated(keep='first') #agregar columna para identificar duplicados

total_por_hotel = hoteles_dup['hotel'].value_counts() #total de reservcas por hotel

sobrantes_por_hotel = hoteles_dup.loc[hoteles_dup['es_duplicado'], 'hotel'].value_counts() ##Filas sobrantes (duplicadas) por hotel
porcentaje_por_hotel = sobrantes_por_hotel / total_por_hotel * 100
print("Sobrantes por hotel")
print(sobrantes_por_hotel)
print()
print("Porcentaje por hotel")
print(round(porcentaje_por_hotel,2))
```

### 2.3 ¿En qué se distinguen las filas duplicadas? (6 pts)

Compara las filas involucradas en duplicados (`keep=False`) con el dataset completo en:

- Distribución de `market_segment` (proporciones).
- Distribución de `deposit_type` (proporciones).
- Tasa de cancelación (`is_canceled`).

```python
# Escribe tu código aquí

completo   = hoteles
duplicados = hoteles[hoteles.duplicated(keep=False)]

def distribucion(df, col):
    return df[col].value_counts(normalize=True).sort_index()

# --- market_segment ---
ms = pd.DataFrame({
    'completo_%'  : (distribucion(completo,   'market_segment') * 100).round(2),
    'duplicados_%': (distribucion(duplicados, 'market_segment') * 100).round(2),
})
ms['dif_pp'] = (ms['duplicados_%'] - ms['completo_%']).round(2)
ms = ms.sort_values('completo_%', ascending=False)
print("market_segment")
print(ms.to_string())

# --- deposit_type ---
dt = pd.DataFrame({
    'completo_%'  : (distribucion(completo,   'deposit_type') * 100).round(2),
    'duplicados_%': (distribucion(duplicados, 'deposit_type') * 100).round(2),
})
dt['dif_pp'] = (dt['duplicados_%'] - dt['completo_%']).round(2)
dt = dt.sort_values('completo_%', ascending=False)
print("deposit_typ")
print(dt.to_string())

# --- is_canceled ---
print("Tasa de cancelación")
print("Completo   :", round(completo['is_canceled'].mean()*100,2), "%")
print("Duplicados :", round(duplicados['is_canceled'].mean()*100,2),"%")
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Escribe tu código aquí
#dataset original
reservas_orig   = hoteles.groupby('hotel').size()
medias_orig     = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean()

metricas_orig = pd.DataFrame({
    'reservas'          : reservas_orig,
    'tasa_cancelacion'  : medias_orig['is_canceled'],
    'adr_promedio'      : medias_orig['adr'],
})

#dataset sin duplicados
reservas_limpio = hoteles_limpio.groupby('hotel').size()
medias_limpio   = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean()

metricas_limpio = pd.DataFrame({
    'reservas'          : reservas_limpio,
    'tasa_cancelacion'  : medias_limpio['is_canceled'],
    'adr_promedio'      : medias_limpio['adr'],
})

print("MÉTRICAS CON DUPLICADOS (hoteles)")
print(metricas_orig.round(2).to_string())
print()
print("MÉTRICAS SIN DUPLICADOS (hoteles_limpio) ")
print(metricas_limpio.round(2).to_string())
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Escribe tu código aquí

#Resumen del dataset ORIGINAL
metricas_orig = (
    hoteles
    .groupby('hotel')
    .agg(
        reservas         = ('is_canceled', 'size'),
        tasa_cancelacion = ('is_canceled', 'mean'),
        adr_promedio     = ('adr', 'mean'),
    )
)
metricas_orig['version'] = 'original'

#Resumen del dataset LIMPIO
metricas_limpio = (
    hoteles_limpio
    .groupby('hotel')
    .agg(
        reservas         = ('is_canceled', 'size'),
        tasa_cancelacion = ('is_canceled', 'mean'),
        adr_promedio     = ('adr', 'mean'),
    )
)
metricas_limpio['version'] = 'sin duplicados'

#Unión con pd.concat()
comparacion = pd.concat([metricas_orig, metricas_limpio])

#Índice renumerado y ordenado por hotel
comparacion = (
    comparacion
    .reset_index()                              # 'hotel' pasa a ser columna
    .sort_values(by=['hotel', 'version'])       # agrupa las dos versiones de cada hotel
    .reset_index(drop=True)                     # renumera el índice desde 0
    .round(4)
)

print("TABLA COMPARATIVA")
print(comparacion.to_string())
```

### 4.2 Duplicados parciales (10 pts)

La columna `reservation_status_date` registra la fecha del último cambio de estado. Detecta las filas de `hoteles_limpio` que **coinciden en todas las columnas excepto `reservation_status_date`**.

1. Cuenta cuántas filas sobrantes hay con ese criterio.
2. Muestra algunas de esas filas ordenadas, con las columnas `hotel`, `lead_time`, `arrival_date_year`, `arrival_date_month`, `arrival_date_day_of_month`, `adr`, `reservation_status` y `reservation_status_date`.
3. Muestra la distribución de `reservation_status` en esas filas.

Después responde: ¿qué podrían representar estas filas? ¿Las eliminarías? Justifica.

```python
# Escribe tu código aquí
#1. Todas las columnas menos reservation_status_date
columnas_clave = [c for c in hoteles_limpio.columns if c != 'reservation_status_date']

#filas que coinciden en todas las columnas clave (keep=False)
mascara_parcial = hoteles_limpio.duplicated(subset=columnas_clave, keep=False)

parciales = hoteles_limpio[mascara_parcial].copy()

sobrantes_parciales = hoteles_limpio.duplicated(subset=columnas_clave, keep='first').sum()
total_involucradas  = mascara_parcial.sum()

print("Filas involucradas en duplicados parciales (keep=False) :", total_involucradas)
print("Filas sobrantes (keep='first')                          :", sobrantes_parciales)

#2. Ordenar filas por x columnas

columnas_ver = [
    'hotel', 'lead_time',
    'arrival_date_year', 'arrival_date_month', 'arrival_date_day_of_month',
    'adr', 'reservation_status', 'reservation_status_date'
]

# Ordenamos por las columnas clave + la fecha, para que las copias queden juntas
parciales_ordenados = parciales.sort_values(
    by=columnas_clave + ['reservation_status_date']
)
print()
print("Primeras 15 filas de duplicados parciales")
print(parciales_ordenados[columnas_ver].head(15).to_string())

#3. Distribución de 'reservation_status'
dist_status = parciales['reservation_status'].value_counts()
dist_status_pct = parciales['reservation_status'].value_counts(normalize=True) * 100
print()
print("Distribución de reservation_status (filas duplicadas parciales)")
print(pd.DataFrame({
    'conteo'    : dist_status,
    'porcentaje': dist_status_pct.round(2)
}).to_string())
```
