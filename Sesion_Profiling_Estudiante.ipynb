## Parte 1 — Funciones a nivel Dataset

Cada función recibe el DataFrame completo (`df`) y regresa **un solo valor**. Completa el cuerpo de cada función donde dice `# TODO`.

```python
def n_columnas(df):
    """Regresa el número de variables (columnas) del DataFrame."""
    columnas = df.shape[1]
    return columnas

def n_filas(df):
    """Regresa el número de observaciones (filas) del DataFrame."""
    filas = df.shape[0]
    return filas
col = n_columnas(df)
fil = n_filas(df)
print('Columnas:',col)
print('Filas:',fil)
```

```python
def n_datos_ausentes(df):
    """Regresa el número total de celdas con datos ausentes (NaN) en todo el DataFrame."""
    return df.isna().sum()

def pct_celdas_ausentes(df):
    """Regresa el porcentaje de celdas con datos ausentes sobre el total de celdas
    (filas x columnas) del DataFrame."""
    return (df.isna().sum().sum()/df.size)*100

print('Datos ausentes:',n_datos_ausentes(df))
print('Porcentaje de celdas ausentes:',round(pct_celdas_ausentes(df),4), '%')
```

```python
def n_filas_duplicadas(df):
    """Regresa el número de filas duplicadas en el DataFrame."""
    return df.duplicated().sum()
    pass

def pct_filas_duplicadas(df):
    """Regresa el porcentaje de filas duplicadas sobre el total de filas."""
    return (df.duplicated().sum().sum()/df.size)*100
    pass

print('Datos duplicados:',n_filas_duplicadas(df))
print('Porcentaje de duplicados:',round(pct_filas_duplicadas(df),4), '%')
```

```python
def n_variables_numericas(df):
    """Regresa el número de columnas con tipo de dato numérico (int o float)."""
    return len(df.select_dtypes(include=['int','float']).columns.tolist())
    pass

def n_variables_booleanas(df):
    """Regresa el número de columnas con tipo de dato booleano."""
    return len(df.select_dtypes(include='bool').columns.tolist())
    pass

def n_variables_categoricas(df):
    """Regresa el número de columnas con tipo de dato object o category."""
    return len(df.select_dtypes(include=['object','category']).columns.tolist())
    pass

print('Columnnas Numericas:',n_variables_numericas(df))
print('Columnas Booleanas:',n_variables_booleanas(df))
print('Columnas Object o Category:',n_variables_categoricas(df))
```

### Función resumen

Combina todas las funciones anteriores en un solo diccionario — tu propio "reporte ejecutivo" del dataset.

```python
def resumen_dataset(df):
    """Regresa un diccionario con el resumen general del DataFrame,
    combinando todas las funciones de la Parte 1."""
    Resumen = {
        'Columnas' : col,
        'Filas': fil,
        'Datos asuentes': len(n_datos_ausentes(df)),
        'Porcentaje de celdas ausentes': round(pct_celdas_ausentes(df),4),
        'Datos duplicados': n_filas_duplicadas(df),
        'Porcentaje duplicados': round(pct_filas_duplicadas(df),4),
        'Columnas Numericas': n_variables_numericas(df),
        'Columnas Booleanas': n_variables_booleanas(df),
        'Columnas Object o Category': n_variables_categoricas(df)
      }
    return Resumen
    pass

print(resumen_dataset(df))
```

## Parte 2 — Funciones a nivel Columna

Cada función recibe el DataFrame y el **nombre de una columna** (`col`), y regresa un solo valor para esa columna.

```python
def total_ausentes(df, col):
    """Regresa el número de valores ausentes (NaN) en la columna col."""
    total = df[col].isna().sum()
    return total
    pass

def pct_ausentes(df, col):
    """Regresa el porcentaje de valores ausentes en la columna col."""
    return (df[col].isna().sum()/df.shape[0])*100
    pass

print('Ausentes:',total_ausentes(df,'Income'))
print('Porcentaje Ausentes:',round(pct_ausentes(df,'Income'),4),'%')
```

```python
def media(df, col):
    """Regresa la media de la columna col (asume que es numérica)."""
    return df[col].mean()
    pass

def minimo(df, col):
    """Regresa el valor mínimo de la columna col."""
    return df[col].min()
    pass

def maximo(df, col):
    """Regresa el valor máximo de la columna col."""
    return df[col].max()
    # TODO
    pass

print('Media:',media(df,'Income'))
print('Minimo:',minimo(df,'Year_Birth'))
print('Maximo:',maximo(df,'Year_Birth'))
```

```python
def total_ceros(df, col):
    """Regresa el número de valores iguales a 0 en la columna col."""
    return (df[col] == 0).sum()
    pass

def pct_ceros(df, col):
    """Regresa el porcentaje de valores iguales a 0 en la columna col."""
    return ((df[col] == 0).sum()/df[col].size)*100
    pass

print('Total ceros:',total_ceros(df,'Complain'))
print('Porcentaje de valores = ceros:',round(pct_ceros(df,'Complain'),4),'%')
```

```python
def total_negativos(df, col):
    """Regresa el número de valores negativos en la columna col."""
    return (df[col] < 0).sum()
    pass

def pct_negativos(df, col):
    """Regresa el porcentaje de valores negativos en la columna col."""
    return ((df[col] < 0).sum()/df[col].size)*100
    pass

print('Valores negativos:',total_negativos(df,'AcceptedCmp3'))
print('Porcentaje Valores negativos:',round(pct_negativos(df,'AcceptedCmp3'),4),'%')
```
