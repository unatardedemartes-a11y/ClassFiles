# 1. Operaciones basicas



Objetivo: Crear y manipular arrays.
Instrucciones:
1. Crea un array de 10 elementos con valores entre 1 y 10.
2. Calcula la media, mediana, y desviacion estandar de los valores.
3. Multiplica cada valor por 2.

```python
import random
import statistics

# Crear un array de 10 elementos con valores entre 1 y 10
array = [random.randint(1, 10) for _ in range(10)]

print("Array original:", array)

# Calcular la media
media = statistics.mean(array)
print("Media:", media)

# Calcular la mediana
mediana = statistics.median(array)
print("Mediana:", mediana)

# Calcular la desviacion estandar
desviacion_estandar = statistics.stdev(array)
print("Desviación estandar:", desviacion_estandar)

# Multiplicar cada valor por 2
array_multiplicado = [x * 2 for x in array]
print("Array multiplicado por 2:", array_multiplicado)
```

# 2. Operaciones con DataFrames

Objetivo: Crear y manipular un DataFrame.

Instrucciones:
1. Crea un DataFrame con 3 columnas (A, B, C) y 5 filas con numeros
aleatorios enteros.
2. Renombra las columnas a Col1, Col2, Col3.
3. Anade una nueva columna que sea la suma de las columnas Col1 y Col2.
4. Elimina todas las filas donde el valor de Col3 sea menor que 10.

```python
import pandas as pd
import numpy as np

# 1. Crear un DataFrame con 3 columnas (A, B, C) y 5 filas con numeros aleatorios enteros
df = pd.DataFrame(np.random.randint(1, 20, size=(5, 3)), columns=['A', 'B', 'C'])
print("DataFrame original:")
print(df)
print()

# 2. Renombrar las columnas a Col1, Col2, Col3
df = df.rename(columns={'A': 'Col1', 'B': 'Col2', 'C': 'Col3'})
print("DataFrame con columnas renombradas:")
print(df)
print()

# 3. Añadir una nueva columna que sea la suma de Col1 y Col2
df['Suma'] = df['Col1'] + df['Col2']
print("DataFrame con la columna 'Suma':")
print(df)
print()

# 4. Eliminar todas las filas donde el valor de Col3 sea menor que 10
df = df[df['Col3'] >= 10]
print("DataFrame final (filas con Col3 >= 10):")
print(df)
```

# 3. Estadıstica descriptiva

Objetivo: Aplicar analisis estadıstico basico.

Instrucciones:
1. Crea un DataFrame con 100 valores aleatorios entre 0 y 50.
2. Calcula la media, mediana y moda de los valores.
3. Encuentra el valor minimo y el valor Maximo del DataFrame.

```python
import pandas as pd
import numpy as np

# Crear DataFrame con 100 valores aleatorios entre 0 y 50
df = pd.DataFrame({'Valores': np.random.randint(0, 51, 100)})

print(df['Valores'])

# Calcular la media
media = statistics.mean(df['Valores'])
print("Media:", media)

# Calcular la mediana
mediana = statistics.median(df['Valores'])
print("Mediana:", mediana)

#Minimo y Maximo
minimo = min(df['Valores'])
maximo = max(df['Valores'])
print("Minimo:", minimo)
print("Maximo:", maximo)
```

# 4. Analisis de series temporales

Objetivo: Trabajar con fechas y series temporales.

Instrucciones:
1. Crea un DataFrame con una columna de fechas (usa pd.date_range()
para generar las fechas).
2. Genera 100 valores aleatorios asociados a estas fechas.
3. Calcula la media movil de los valores con una ventana de 5 dıas.

```python
import pandas as pd
import numpy as np

#Crear dataframe con columna de fechas
fechas = pd.date_range(start='2026-01-01', periods=100, freq='D')

#Gnerar 100 valores aleatorios
valores = np.random.uniform(0, 50, 100)

df = pd.DataFrame({
    'Fecha': fechas,
    'Valor': valores
})
print(df.head(10))
print()

#Calcular la media movilcon una ventana de 5 dias
df['Media_Movil_5'] = df['Valor'].rolling(window=5).mean()
print("DataFrame con la media movil:")
print(df.head(15))
```

# 5. Grafico de Lıneas

Objetivo: Visualizar la tendencia de una serie de datos.

Instrucciones:

* Crea una serie de datos aleatorios (entre 0 y 100) de tamano 50.
* Usa Matplotlib para graficar los datos en un grafico de lıneas.
* Anade tıtulo, etiquetas a los ejes y una leyenda.

```python

import matplotlib.pyplot as plt
import numpy as np

#Crear serie de datos aleatorios entre 0 y 100 de tamano 50
datos = np.random.uniform(0, 100, 50)

#Crear grafico de lineas
plt.plot(datos)
plt.show()

#Anadir titulo y etiquetas a los ejes
plt.title('Grafico de Lineas - Serie de Datos Aleatorios', fontsize=14, fontweight='bold')
plt.xlabel('Indice', fontsize=12)
plt.ylabel('Valor', fontsize=12)

#Anadir leyenda
plt.legend(['Datos aleatorios'])
plt.legend(loc='upper right', fontsize=11)

#Cambiar el color y estilo de la linea
plt.plot(datos, color='red', linestyle='--', linewidth=2, label='Datos aleatorios')
plt.show()
```

# 6. Grafico de Barras

Objetivo: Visualizar datos categoricos.

Instrucciones:
1. Carga el dataset tips de Seaborn.
2. Realiza un grafico de barras que muestre el total de las cuentas (total_bill)
agrupadas por dıa.
3. Anade etiquetas a los ejes y un tıtulo.


```python
import seaborn as sns
import matplotlib.pyplot as plt

#Carga el dataset tips de Seaborn
tips = sns.load_dataset('tips')
print("Primeras filas del dataset:")
print(tips.head())
print()
print("Forma:", tips.shape)

# Agrupar por dia y sumar total_bill
totales_por_dia = tips.groupby('day', observed=True)['total_bill'].sum()

# Crear grafico de barras
plt.bar(totales_por_dia.index, totales_por_dia.values, color='steelblue')

#Anadir etiquetas y ejes
plt.title('Total de Cuentas por Dia', fontsize=14, fontweight='bold')
plt.xlabel('Dia de la semana', fontsize=12)
plt.ylabel('Total de cuentas (total_bill)', fontsize=12)
plt.show()
```

# 7. Histograma

Objetivo: Visualizar la distribucion de datos numericos.

Instrucciones:
1. Genera 1000 numeros aleatorios con distribucion normal (media = 0, desviacion estandar = 1).
2. Crea un histograma para visualizar la distribucion.
3. Ajusta el numero de bins del histograma para que la visualizacion sea
clara.

```python
import numpy as np
import matplotlib.pyplot as plt

#Generar 1000 numeros aleatorios con distribucion normal
datos = np.random.normal(loc=0, scale=1, size=1000)

#Verificamos
print("Verificacion")
print("Media:", datos.mean())
print("Desviacion muestral:", datos.std())
print("Tamano:", len(datos))

#Crear histograma
plt.hist(datos)
plt.show()

#Ajustar el numero de bins
plt.hist(datos, bins=30, color='steelblue', edgecolor='black', alpha=0.75)

# Añadir título y etiquetas
plt.title('Histograma - Distribución Normal (μ=0, σ=1, n=1000)',
          fontsize=14, fontweight='bold')
plt.xlabel('Valor', fontsize=12)
plt.ylabel('Frecuencia', fontsize=12)
plt.show()
```

# 8. Scatter Plot


Objetivo: Visualizar la relacion entre dos variables.

Instrucciones:
1. Usa el dataset iris de Seaborn.
2. Crea un grafico de dispersion (scatter plot) para visualizar la relacion entre el ancho del sepalo (sepal_width) y la longitud del sepalo
(sepal_length).
3. Anade colores diferentes segun la especie de flor.

```python
import seaborn as sns
import matplotlib.pyplot as plt

#Cargar dataset
iris = sns.load_dataset('iris')
print("Primeras filas del dataset:")
print(iris.head())
print()
print("Forma:", iris.shape)

#Crear grafico
plt.scatter(iris['sepal_width'], iris['sepal_length'])
plt.show()

#Anadir colores
colores = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}

#Graficar cada especie con su color
for especie, color in colores.items():
    subset = iris[iris['species'] == especie]
    plt.scatter(subset['sepal_width'],
                subset['sepal_length'],
                color=color,
                label=especie,
                alpha=0.7,
                s=60,
                edgecolor='black')

#titulos y etiquetas
plt.title('Relación entre Ancho y Longitud del Sépalo por Especie',
          fontsize=14, fontweight='bold')
plt.xlabel('Ancho del Sépalo (sepal_width)', fontsize=12)
plt.ylabel('Longitud del Sépalo (sepal_length)', fontsize=12)

plt.legend(title='Especie', fontsize=11)
plt.show()
```

# 9. Heatmap

Objetivo: Visualizar la correlacion entre variables numericas.

Instrucciones:
1. Usa el dataset iris de Seaborn.

```python


#Crear matriz correlacion usando el dataset creado en ejercicio 8
matriz_corr = iris.select_dtypes(include=[np.number]).corr()
print("Matriz de correlación:")
print(matriz_corr.round(3))

#Crear heatmap de la correlacion
sns.heatmap(matriz_corr,
            annot=True,
            cmap='coolwarm',
            fmt='.2f',
            vmin=-1, vmax=1,
            square=True,
            linewidths=0.5,
            cbar_kws={'label': 'Coeficiente de correlación'})

#Titulo
plt.title('Matriz de Correlación - Dataset Iris',
          fontsize=14, fontweight='bold')
plt.show()
```

# 11.  Datos Temporales con Analisis Estadıstico


Objetivo: Trabajar con series temporales y calcular estadısticas.
Instrucciones:
1. Genera una serie temporal de 200 dıas con valores aleatorios.
2. Calcula la media y la desviacion estandar de la serie temporal.
3. Visualiza la serie junto con las bandas de desviacion estandar (es decir, la
media ± 1 desviacion estandar).

```python
#Generar un serie temporal de 200 dias con valores aleatorios
fechas_200 = pd.date_range(start='2026-01-01', periods=200, freq='D')
valores_200 = np.random.uniform(0, 100, 200)
df_200 = pd.DataFrame({'Valor': valores_200}, index=fechas_200)
df_200.index.name = 'Fecha'

#Media y desviación stdr
media = df_200['Valor'].mean()
std   = df_200['Valor'].std()
print("Media:", media)
print("Desviación estándar:",std)

#Grafico con bandas +- 1σ
#Serie temporal
plt.plot(df_200.index, df_200['Valor'], color='steelblue', label='Serie')

#Media
plt.axhline(media, color='green', linestyle='--', label=f'Media = {media:.2f}')

#Bandas +_1σ
plt.axhline(media + std, color='red', linestyle=':', linewidth=1.5)
plt.axhline(media - std, color='red', linestyle=':', linewidth=1.5)
plt.fill_between(df_200.index, media - std, media + std,
                 color='red', alpha=0.15, label='Banda +_1σ')

#Titulo y etiquetas
plt.title('Serie Temporal con Media y Bandas +_1σ', fontsize=14, fontweight='bold')
plt.xlabel('Fecha')
plt.ylabel('Valor')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
