### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# Escribe tu código aquí

#crear la columna 'duracion' como diferencia entre dropoff y pickup
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']

#convertir la duracion a minutos
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60

#resumen estadistico de duracion_min
print(taxis['duracion_min'].describe())
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# Escribe tu código aquí\
# calcular 'velocidad_mph' solo para viajes con duracion > 0
taxis['velocidad_mph'] = (taxis['distance'] / (taxis['duracion_min'] / 60)).where(
    taxis['duracion_min'] > 0
)

#contar viajes con velocidad > 60 mph
mayor_60 = (taxis['velocidad_mph'] > 60).sum()
print("Viajes con velocidad mayor a 60 mph:", mayor_60)

#tpo 5 de viajes con mayor velocidad
columnas = ['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']
top5 = taxis.nlargest(5, 'velocidad_mph')[columnas]
print()
print("Top 5 viajes con mayor velocidad:")
print(top5)
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Escribe tu código aquí
#hora de pickup
taxis['hora'] = taxis['pickup'].dt.hour

#dia de la semana de pickup (lunes = 0)
taxis['dia_num'] = taxis['pickup'].dt.dayofweek

#nombre del dia de pickup
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()

#fin de semana
taxis['fin_semana'] = taxis['dia_num'].isin([5, 6])

#solo la fecha sin hora de pickup
taxis['fecha'] = taxis['pickup'].dt.date

#fecha de dropoff distinta de la de pickup
taxis['cruza_medianoche'] = taxis['pickup'].dt.date != taxis['dropoff'].dt.date

#contar viajes que cruzan medianoche
cruzan = taxis['cruza_medianoche'].sum()
print("Viajes que cruzan la medianoche:", cruzan)
print()
print(taxis)
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# Escribe tu código aquí
#agrupar por hora y calcular numero de viajes y mediana de duracion
por_hora = taxis.groupby('hora').agg(
    num_viajes=('duracion_min', 'size'),
    mediana_duracion=('duracion_min', 'median')
)

print("Patron por hora del dia:")
print(por_hora)
print()

#hora con mas viajes
hora_max = por_hora['num_viajes'].idxmax()
print("Hora con mas viajes:", str(hora_max).zfill(2) + ":00", por_hora.loc[hora_max, 'num_viajes'], "viajes")

# Hora con menos viajes
hora_min = por_hora['num_viajes'].idxmin()
print("Hora con menos viajes:", str(hora_min).zfill(2) + ":00",por_hora.loc[hora_min, 'num_viajes'], "viajes")
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Escribe tu código aquí

#agrupacion principal
por_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    num_viajes=('duracion_min', 'size'),
    mediana_duracion=('duracion_min', 'median'),
    mediana_fare=('fare', 'median')
)

#numero de fechas distintas por dia de la semana
fechas_por_dia = taxis.groupby(['dia_num', 'dia_nombre'])['fecha'].nunique()

#anadir columna de promedio
por_dia['fechas_distintas'] = fechas_por_dia
por_dia['promedio_viajes_por_fecha'] = (por_dia['num_viajes'] / por_dia['fechas_distintas']).round(2)

print("Patron por dia de la semana")
print(por_dia)
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Escribe tu código aquí
#definir cada condición por separado
cond_periodo    = (taxis['pickup'].dt.year == 2019) & (taxis['pickup'].dt.month == 3)
cond_duracion   = taxis['duracion_min'] >= 1
cond_velocidad  = (taxis['velocidad_mph'] > 0) & (taxis['velocidad_mph'] <= 60)
cond_distancia  = taxis['distance'] > 0

#contar exclusiones por condicion
print("Viajes excluidos por condicion")
print("Fuera de marzo 2019        :", (~cond_periodo).sum())
print("Duración < 1 minuto        :", (~cond_duracion).sum())
print("Velocidad fuera de (0, 60] :", (~cond_velocidad).sum())
print("Distancia <= 0             :", (~cond_distancia).sum())
print()

#mascara combinada
mascara = cond_periodo & cond_duracion & cond_velocidad & cond_distancia

#crear taxis_limpio
taxis_limpio = taxis[mascara].copy()

print("Total de viajes originales :", len(taxis))
print("Total de viajes excluidos  :", (~mascara).sum())
print("Total de viajes en limpio  :", len(taxis_limpio))
print()

#verificacion
print("Verificacion")
print("Todos en marzo 2019        :", ((taxis_limpio['pickup'].dt.year == 2019) & (taxis_limpio['pickup'].dt.month == 3)).all())
print("Todos con duracion >= 1    :", (taxis_limpio['duracion_min'] >= 1).all())
print("Todos con velocidad en (0,60]:", ((taxis_limpio['velocidad_mph'] > 0) & (taxis_limpio['velocidad_mph'] <= 60)).all())
print("Todos con distance > 0     :", (taxis_limpio['distance'] > 0).all())
```

```python
# Escribe tu código aquí
#filtrar horario 7-9
taxis_horario = taxis_limpio[taxis_limpio['hora'].between(7, 9)]

#agrupar por fin_semana
resumen = taxis_horario.groupby('fin_semana').agg(
    num_viajes=('duracion_min', 'size'),
    mediana_duracion=('duracion_min', 'median'),
    mediana_velocidad=('velocidad_mph', 'median')
)

print(resumen)
```
