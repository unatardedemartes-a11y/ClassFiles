```python
import numpy as np
import matplotlib.pyplot as plt

# --- Caso 1: MNIST ---
N1, t01, c1 = 10**4, 0.01, 10**-6
B1 = np.linspace(10, 1500, 500)
T1 = (N1 * t01) / B1 + c1 * B1**2
B_opt1 = (N1 * t01 / (2 * c1))**(1/3)
T_opt1 = (N1 * t01) / B_opt1 + c1 * B_opt1**2

# --- Caso 2: ImageNet ---
N2, t02, c2 = 10**8, 2, 10**-4
B2 = np.linspace(1000, 30000, 500)
T2 = (N2 * t02) / B2 + c2 * B2**2
B_opt2 = (N2 * t02 / (2 * c2))**(1/3)
T_opt2 = (N2 * t02) / B_opt2 + c2 * B_opt2**2

# --- Graficar ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1
ax1.plot(B1, T1, label='T(B)', color='blue')
ax1.scatter([B_opt1], [T_opt1], color='red', zorder=5, label=f'Mínimo: B={B_opt1:.0f}')
ax1.set_title('Caso 1: MNIST (CPU)')
ax1.set_xlabel('Tamaño de Lote (B)')
ax1.set_ylabel('Tiempo Total (s)')
ax1.grid(True, linestyle='--', alpha=0.7)
ax1.legend()

# Gráfico 2
ax2.plot(B2, T2, label='T(B)', color='green')
ax2.scatter([B_opt2], [T_opt2], color='red', zorder=5, label=f'Mínimo: B={B_opt2:.0f}')
ax2.set_title('Caso 2: ImageNet (GPU)')
ax2.set_xlabel('Tamaño de Lote (B)')
ax2.set_ylabel('Tiempo Total (s)')
ax2.grid(True, linestyle='--', alpha=0.7)
ax2.legend()

plt.tight_layout()
plt.show()
```
